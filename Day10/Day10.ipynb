```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E",
                "F", "G", "H", "I", "J"],
    "Study_Hours": [1, 2, 2, 3, 3, 4, 4, 5, 6, 10],
    "Exam_Score": [35, 40, 42, 50, 50, 60, 65, 70, 75, 95]
}

df = pd.DataFrame(data)

display(df)
```

```python
scores = df["Exam_Score"]

print("Mean:", scores.mean())
print("Median:", scores.median())
print("Mode:", scores.mode().tolist())
```

```python
print("Sample Variance:", scores.var())
print("Sample Standard Deviation:", scores.std())

print("Population Variance:", scores.var(ddof=0))
print("Population Standard Deviation:", scores.std(ddof=0))
```

```python
print("Q1:", scores.quantile(0.25))
print("Q2:", scores.quantile(0.50))
print("Q3:", scores.quantile(0.75))
print("90th Percentile:", scores.quantile(0.90))
```

```python
q1 = scores.quantile(0.25)
q3 = scores.quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = scores[
    (scores < lower_bound) |
    (scores > upper_bound)
]

print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers.tolist())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Exam_Score",
    bins=6,
    kde=True
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["Exam_Score"])

plt.title("Box Plot of Exam Scores")
plt.xlabel("Exam Score")

plt.tight_layout()
plt.show()
```

```python
normal_scores = [40, 45, 50, 55, 60]
scores_with_outlier = [40, 45, 50, 55, 200]

comparison = pd.DataFrame({
    "Dataset": ["Original", "With Outlier"],
    "Mean": [
        np.mean(normal_scores),
        np.mean(scores_with_outlier)
    ],
    "Median": [
        np.median(normal_scores),
        np.median(scores_with_outlier)
    ]
})

display(comparison)
```

```python
correlation = df["Study_Hours"].corr(df["Exam_Score"])

print("Correlation:", round(correlation, 3))
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Study_Hours",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import os

for root, dirs, files in os.walk("../data/raw/archive"):
    for file in files:
        if file.endswith(".csv"):
            file_path = os.path.join(root, file)

retail_data = pd.read_csv(file_path, encoding="latin1")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", retail_data.shape)

print("\nColumn Names:")
print(retail_data.columns.tolist())

print("\nFirst 5 Records:")
display(retail_data.head())
```

```python
print("Dataset Shape:", retail_data.shape)

print("\nDataset Information:")
retail_data.info()

print("\nStatistical Summary:")
display(retail_data.describe())

print("\nMissing Values:")
display(retail_data.isnull().sum())

print("\nDuplicate Records:")
print(retail_data.duplicated().sum())
```

```python
print("Missing Values:")
print(retail_data.isnull().sum())

print("\nTotal Missing Values:")
print(retail_data.isnull().sum().sum())

print("\nDuplicate Rows:")
print(retail_data.duplicated().sum())

print("\nUnique Customers:")
print(retail_data["Customer ID"].nunique())

print("\nUnique Orders:")
print(retail_data["Order ID"].nunique())

print("\nUnique Products:")
print(retail_data["Product ID"].nunique())
```

```python
retail_data["Order Date"] = pd.to_datetime(
    retail_data["Order Date"],
    format="%m/%d/%Y"
)

retail_data["Ship Date"] = pd.to_datetime(
    retail_data["Ship Date"],
    format="%m/%d/%Y"
)

print("Date Conversion Completed!")

print("\nUpdated Data Types:")
print(retail_data[["Order Date", "Ship Date"]].dtypes)

display(retail_data[["Order Date", "Ship Date"]].head())
```

```python
retail_data["Order Date"] = pd.to_datetime(
    retail_data["Order Date"],
    format="%m/%d/%Y"
)

retail_data["Ship Date"] = pd.to_datetime(
    retail_data["Ship Date"],
    format="%m/%d/%Y"
)

print("Date Conversion Completed!")

print("\nUpdated Data Types:")
print(retail_data[["Order Date", "Ship Date"]].dtypes)

display(retail_data[["Order Date", "Ship Date"]].head())
```

```python
retail_data["Year"] = retail_data["Order Date"].dt.year

retail_data["Month"] = retail_data["Order Date"].dt.month

retail_data["Month Name"] = retail_data["Order Date"].dt.month_name()

retail_data["Shipping Days"] = (
    retail_data["Ship Date"] - retail_data["Order Date"]
).dt.days

print("New Columns Created Successfully!")

display(
    retail_data[
        ["Order Date", "Year", "Month", "Month Name", "Shipping Days"]
    ].head()
)
```

```python
total_sales = retail_data["Sales"].sum()
total_profit = retail_data["Profit"].sum()
total_orders = retail_data["Order ID"].nunique()
total_customers = retail_data["Customer ID"].nunique()
total_products = retail_data["Product ID"].nunique()
total_quantity = retail_data["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100
average_order_value = total_sales / total_orders

print("===== RETAIL BUSINESS KPIs =====")

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products:", total_products)
print("Total Quantity Sold:", total_quantity)
print("Profit Margin:", round(profit_margin, 2), "%")
print("Average Order Value:", round(average_order_value, 2))
```

```python
region_analysis = retail_data.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

region_analysis = region_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

display(region_analysis)
```

```python
category_analysis = retail_data.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()

category_analysis = category_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

display(category_analysis)
```

```python
subcategory_analysis = retail_data.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

subcategory_analysis = subcategory_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

display(subcategory_analysis)
```

```python
customer_analysis = retail_data.groupby(
    ["Customer ID", "Customer Name"]
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

customer_analysis = customer_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("Top 10 Customers by Sales:")

display(customer_analysis.head(10))
```

```python
monthly_sales = retail_data.groupby(
    ["Year", "Month", "Month Name"]
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()

monthly_sales = monthly_sales.sort_values(
    by=["Year", "Month"]
)

display(monthly_sales)
```

```python
monthly_sales = retail_data.groupby(
    ["Year", "Month", "Month Name"]
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()

monthly_sales = monthly_sales.sort_values(
    by=["Year", "Month"]
)

display(monthly_sales)
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="Month Name",
    y="Total_Sales",
    hue="Year",
    marker="o"
)

plt.title("Monthly Sales Trend Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.legend(title="Year")
plt.show()
```

```python
product_analysis = retail_data.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

top_10_products = product_analysis.sort_values(
    by="Total_Sales",
    ascending=False
).head(10)

display(top_10_products)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_10_products,
    x="Total_Sales",
    y="Product Name"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.grid(axis="x")
plt.show()
```

```python
top_profitable_products = product_analysis.sort_values(
    by="Total_Profit",
    ascending=False
).head(10)

display(top_profitable_products)
```

```python
loss_products = product_analysis[
    product_analysis["Total_Profit"] < 0
].sort_values(
    by="Total_Profit",
    ascending=True
).head(10)

display(loss_products)
```

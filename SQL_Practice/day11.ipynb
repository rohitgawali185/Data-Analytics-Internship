```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Profit": [2000, 3000, 4000, 2500, 5000, 6000],
    "Customers": [100, 120, 150, 110, 180, 200],
    "Advertising": [1000, 1500, 2000, 1200, 2500, 3000]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Sales"], label="Sales")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=5, edgecolor="black")

plt.title("Sales Distribution")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(y=df["Sales"])

plt.title("Sales Outlier Analysis")
plt.ylabel("Sales Amount")
plt.tight_layout()
plt.show()
```

```python
students = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Marks": [75, 45, 88, 32]
}

for j in range(len(students["Name"])):

    name = students["Name"][j]
    marks = students["Marks"][j]

    if marks >= 40:
        result = "Pass"
    else:
        result = "Fail"

    print("Name:", name)
    print("Marks:", marks)
    print("Result:", result)
    print("----------------")
    
```

```python
def calculate_salary(monthly_salary):
    annual_salary = monthly_salary * 12
    
    return annual_salary
```

```python
result = calculate_salary(25000)

print("Annual Salary:", result)
```

```python
salaries = [25000, 30000, 35000]

total_salary = sum(salaries)
total_employees = len(salaries)

average_salary = total_salary / total_employees

print("Total Salary:", total_salary)
print("Total Employees:", total_employees)
print("Average Monthly Salary:", average_salary)
```

```python
def calculate_annual_salary(monthly_salary):
    return monthly_salary * 12


def calculate_average_salary(salaries):
    total_salary = sum(salaries)
    total_employees = len(salaries)
    return total_salary / total_employees


# Employee salaries
salaries = [25000, 30000, 35000]

print("Annual Salary:", calculate_annual_salary(25000))
print("Average Monthly Salary:", calculate_average_salary(salaries))
```

```python
%%sql

CREATE TABLE Employees (
    EmpID INT PRIMARY KEY,
    Name VARCHAR(50),
    Department VARCHAR(50),
    Salary INT,
    Age INT
);
```

```python
%%sql

INSERT INTO Employees VALUES
(1, 'Amit', 'HR', 30000, 25),
(2, 'Rahul', 'IT', 45000, 28),
(3, 'Sneha', 'Finance', 35000, 24),
(4, 'Priya', 'IT', 50000, 30),
(5, 'Neha', 'HR', 28000, 26);
```

```python
%%sql

SELECT * FROM Employees;
```

```python
%%sql

SELECT *
FROM Employees
WHERE Department = 'IT';
```

```python
%%sql

SELECT *
FROM Employees
ORDER BY Salary DESC;
```

```python
%%sql

SELECT Department, COUNT(*) AS Total_Employees
FROM Employees
GROUP BY Department;
```

```python
%%sql

SELECT
    SUM(Salary) AS Total_Salary,
    AVG(Salary) AS Average_Salary,
    MAX(Salary) AS Highest_Salary,
    MIN(Salary) AS Lowest_Salary
FROM Employees;
```
